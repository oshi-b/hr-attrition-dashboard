```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("D:\HR Analytics Project\excel\WA_Fn-UseC_-HR-Employee-Attrition.csv")
print(df.shape)        # (1470, 35)
print(df.dtypes)
print(df.isnull().sum())  # should be 0 nulls
print(df['Attrition'].value_counts(normalize=True).round(2))
# No = 0.84, Yes = 0.16
```

```python
from sklearn.preprocessing import LabelEncoder

df_clean = df.drop(columns=['EmployeeCount','Over18','StandardHours','EmployeeNumber'])

# Encode categorical columns
cat_cols = df_clean.select_dtypes(include='object').columns.tolist()
cat_cols.remove('Attrition')

le = LabelEncoder()
for col in cat_cols:
    df_clean[col + '_enc'] = le.fit_transform(df_clean[col])

df_clean.to_csv('../data/processed/hr_cleaned.csv', index=False)
print(f"Saved: {df_clean.shape}")
```
